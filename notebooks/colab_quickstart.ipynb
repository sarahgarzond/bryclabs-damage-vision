Runtime → Change runtime type → **GPU**. Runs Qwen2.5-VL-7B in 4-bit with the `hf` backend.

```
!git clone https://github.com/sarahgarzond/bryclabs-damage-vision
%cd bryclabs-damage-vision
!pip install -q -r requirements-gpu.txt
```

Upload a few images into `data/images/` (left sidebar → Files), or mount Google Drive and copy them.

```
import torch; print(torch.cuda.get_device_name(0))
!ls data/images | head
```

Run on 3 images first. `--no-verify` skips the detector for speed. Remove it once outputs look right.

```
!python scripts/run.py --images data/images --out outputs/colab --backend hf --load-4bit --runs 1 --no-verify --limit 3
```

```
import json, glob
for p in sorted(glob.glob('outputs/colab/*.json'))[:3]:
    d = json.load(open(p))
    print(p, d['status'], [(f['damage_type'], f['component'], f['confidence']) for f in d['findings']])
    print('  limitations:', d['limitations'])
```
